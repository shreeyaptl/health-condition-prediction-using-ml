Handle Numerical Missing Values in the Dataset

Handle Categorical Missing Values in the Dataset

```python
# replace the missing value with mode for ordinal columns
ordinal_cols = [
    "stress_level",
    "sleep_quality",
    "physical_activity_level",
    "smoking_alcohol"
]

for col in ordinal_cols:
    df_new[col] = df_new[col].fillna(df_new[col].mode()[0])


# impute values for diet_type and gender columns

# Fill using the first mode value found


df_new['diet_type'] = df_new['diet_type'].fillna(df_new['diet_type'].mode()[0])
df_new['gender'] = df_new['gender'].fillna(df_new['gender'].mode()[0])
```

ORDINAL ENCODING FOR CATEGORICAL COLUMNS

```python
import pandas as pd
from sklearn.preprocessing import OrdinalEncoder
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer

# 1. ORDINAL ENCODING
# Create mappings

smoking_mapping = {
    "no": 0,
    "occasional": 1,
    "yes": 2
}

physical_activity_mapping = {
    "sedentary": 0,
    "moderate": 1,
    "active": 2
}

sleep_quality_mapping = {
    "poor": 0,
    "average": 1,
    "good": 2
}

stress_level_mapping = {
    "low": 0,
    "medium": 1,
    "high": 2
}

# Define the columns to be ordinal encoded
columns_to_encode = ["smoking_alcohol", "physical_activity_level", "sleep_quality", "stress_level"]

for col in columns_to_encode:
    # Revert to original string values from 'df' (which should hold the raw categories)
    df_new[col] = df[col]
    # Fill any NaNs using the mode of the current df_new column (which now contains original strings)
    df_new[col] = df_new[col].fillna(df_new[col].mode()[0])

df_new[columns_to_encode] = df_new[columns_to_encode].astype(str)

# Ordinal encoder
# Convert dictionaries of mapping to list
# the encoder receives list of lists
encoder = OrdinalEncoder(categories = [list(smoking_mapping.keys()), list(physical_activity_mapping.keys()), list(sleep_quality_mapping.keys()), list(stress_level_mapping.keys())])

df_new[columns_to_encode] = encoder.fit_transform(df_new[columns_to_encode])

#print the first few values in the column
print(df_new.head())
```

ONE-HOT ENCODING FOR CATEGORICAL COLUMNS

```python
# 2. ONE-HOT ENCODING


nominal_columns = [
    "gender",
    "diet_type"
]


preprocessor = ColumnTransformer(
    transformers=[
        (
            "onehot",
            OneHotEncoder(
                sparse_output=False,
                handle_unknown="ignore"
            ),
            nominal_columns
        )
    ],
    remainder="passthrough"
)



# 3. APPLY TRANSFORMATION

encoded_array = preprocessor.fit_transform(df_new)



# 4. CREATE FINAL DATAFRAME

encoded_columns = preprocessor.get_feature_names_out()

encoded_df = pd.DataFrame(
    encoded_array,
    columns=encoded_columns,
    index=df_new.index
)


print(encoded_df.head())

print("\nShape of encoded dataframe:")
print(encoded_df.shape)
```

```python
#Implementing the train and test set
X = encoded_df.drop(columns = ['remainder__id','remainder__health_condition'])
Y = encoded_df['remainder__health_condition']


from sklearn.model_selection import train_test_split

X_train, X_valid, Y_train, Y_valid = train_test_split(
    X,
    Y,
    test_size=0.2,
    random_state=42,
    stratify=Y
)

# Print shapes using .shape attribute (not a method)
print("x_train shape:", X_train.shape)
print("y_train shape:", Y_train.shape)


#Implement a random forest classifier
from sklearn.ensemble import RandomForestClassifier

model = RandomForestClassifier(
    random_state=42,
    n_estimators=200
)

model.fit(X_train, Y_train)
```

```python

from sklearn.metrics import balanced_accuracy_score

Y_pred = model.predict(X_valid)

score = balanced_accuracy_score(Y_valid, Y_pred)

print(score)
```

```python
from sklearn.metrics import classification_report, confusion_matrix

print(classification_report(Y_valid, Y_pred))
print(confusion_matrix(Y_valid, Y_pred))
```

```python
Y_test = test_df.copy()

# 1. Handle Numerical Missing Values (using means from df_new, which was already imputed)
Y_test['sleep_duration'] = Y_test['sleep_duration'].fillna(df_new['sleep_duration'].mean())
Y_test['heart_rate'] = Y_test['heart_rate'].fillna(df_new['heart_rate'].mean())
Y_test['bmi'] = Y_test['bmi'].fillna(df_new['bmi'].mean())
Y_test['calorie_expenditure'] = Y_test['calorie_expenditure'].fillna(df_new['calorie_expenditure'].mean())
Y_test['step_count'] = Y_test['step_count'].fillna(df_new['step_count'].mean())
Y_test['exercise_duration'] = Y_test['exercise_duration'].fillna(df_new['exercise_duration'].mean())
Y_test['water_intake'] = Y_test['water_intake'].fillna(df_new['water_intake'].mean())

# 2. Handle Categorical Missing Values (using modes from the original training dataframe `df`)
ordinal_cols = ["stress_level", "sleep_quality", "physical_activity_level", "smoking_alcohol"]
for col in ordinal_cols:
    # Use the mode from the original training dataframe (df) to fill NaNs in Y_test
    Y_test[col] = Y_test[col].fillna(df[col].mode()[0])

Y_test['diet_type'] = Y_test['diet_type'].fillna(df['diet_type'].mode()[0])
Y_test['gender'] = Y_test['gender'].fillna(df['gender'].mode()[0])

# 3. Ordinal Encoding (using the already fitted encoder)
columns_to_encode = ["smoking_alcohol", "physical_activity_level", "sleep_quality", "stress_level"]
# Ensure the columns are of string type before ordinal encoding
Y_test[columns_to_encode] = Y_test[columns_to_encode].astype(str)
Y_test[columns_to_encode] = encoder.transform(Y_test[columns_to_encode])

# 4. One-Hot Encoding (using the already fitted preprocessor)
# Add a dummy 'health_condition' column to Y_test to match the columns the preprocessor was fitted on.
# The actual values do not matter as this column will be filtered out from the final features.
Y_test['health_condition'] = 'placeholder'
test_encoded_array = preprocessor.transform(Y_test)

# Get all feature names from the fitted preprocessor
all_encoded_test_feature_names = preprocessor.get_feature_names_out()

# Create a DataFrame with all 19 columns initially
temp_test_encoded_df = pd.DataFrame(
    test_encoded_array,
    columns=all_encoded_test_feature_names,
    index=Y_test.index
)

# 5. Drop the 'id' and 'health_condition' columns to match the training features (X_train)
X_test_processed = temp_test_encoded_df.drop(columns=['remainder__id', 'remainder__health_condition'])

y_test_predictions = model.predict(X_test_processed)
```
